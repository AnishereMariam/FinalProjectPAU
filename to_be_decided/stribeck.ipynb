```python
import React from 'react';
import { LineChart, Line, XAxis, YAxis, CartesianGrid, Tooltip, Legend, ResponsiveContainer, ReferenceLine, Label } from 'recharts';

const StribeckCurve = () => {
  const generateData = () => {
    const data = [];
    const wearFactors = [0, 0.3, 0.6, 0.9];
    
    for (let omega = 10; omega <= 2000; omega += 10) {
      const point = { omega };
      
      wearFactors.forEach(wearFactor => {
        let mu;
        
        if (omega < 100) {
          mu = 0.19 * (1 + wearFactor);
        }
        else if (omega < 1000) {
          mu = 0.11 * (1 + 0.5 * wearFactor);
        }
        else {
          mu = 0.045 * (1 + 0.2 * wearFactor);
        }
        
        const key = 'wear_' + wearFactor.toString().replace('.', '_');
        point[key] = mu;
      });
      
      data.push(point);
    }
    
    return data;
  };

  const data = generateData();

  const CustomTooltip = ({ active, payload, label }) => {
    if (active && payload && payload.length) {
      const rpmValue = Math.round((label * 60) / (2 * Math.PI));
      return (
        <div style={{
          backgroundColor: 'white',
          padding: '10px',
          border: '1px solid #ccc',
          borderRadius: '4px'
        }}>
          <p style={{ margin: '0 0 5px 0', fontWeight: 'bold' }}>
            Angular velocity: {label} rad/s ({rpmValue} rpm)
          </p>
          {payload.map((entry, index) => (
            <p key={index} style={{ margin: '2px 0', color: entry.color }}>
              Friction: {entry.value.toFixed(4)}
            </p>
          ))}
        </div>
      );
    }
    return null;
  };

  return (
    <div style={{ width: '100%', padding: '20px' }}>
      <h2 style={{ textAlign: 'center', marginBottom: '10px' }}>
        Stribeck Curve for Nigerian Tropical Manufacturing Conditions
      </h2>
      <p style={{ textAlign: 'center', color: '#666', marginBottom: '20px' }}>
        Friction Coefficient vs Angular Velocity for Different Wear Levels
      </p>
      
      <ResponsiveContainer width="100%" height={500}>
        <LineChart data={data} margin={{ top: 5, right: 30, left: 20, bottom: 80 }}>
          <CartesianGrid strokeDasharray="3 3" />
          <XAxis 
            dataKey="omega" 
            scale="log" 
            domain={[10, 2000]}
            ticks={[10, 50, 100, 200, 500, 1000, 2000]}
            label={{ value: 'Angular Velocity (rad/s)', position: 'insideBottom', offset: -10 }}
          />
          <YAxis 
            domain={[0, 0.25]}
            label={{ value: 'Friction Coefficient', angle: -90, position: 'insideLeft' }}
          />
          <Tooltip content={<CustomTooltip />} />
          <Legend 
            verticalAlign="top" 
            height={36}
          />
          
          <ReferenceLine x={100} stroke="#999" strokeDasharray="5 5">
            <Label value="Boundary to Mixed" position="top" offset={10} />
          </ReferenceLine>
          <ReferenceLine x={1000} stroke="#999" strokeDasharray="5 5">
            <Label value="Mixed to Hydrodynamic" position="top" offset={10} />
          </ReferenceLine>
          
          <Line 
            type="monotone" 
            dataKey="wear_0" 
            stroke="#2563eb" 
            strokeWidth={2.5}
            dot={false}
            name="No wear (0%)"
          />
          <Line 
            type="monotone" 
            dataKey="wear_0_3" 
            stroke="#16a34a" 
            strokeWidth={2.5}
            dot={false}
            name="Light wear (30%)"
          />
          <Line 
            type="monotone" 
            dataKey="wear_0_6" 
            stroke="#ea580c" 
            strokeWidth={2.5}
            dot={false}
            name="Moderate wear (60%)"
          />
          <Line 
            type="monotone" 
            dataKey="wear_0_9" 
            stroke="#dc2626" 
            strokeWidth={2.5}
            dot={false}
            name="Heavy wear (90%)"
          />
        </LineChart>
      </ResponsiveContainer>

      <div style={{ marginTop: '30px', padding: '15px', backgroundColor: '#f3f4f6', borderRadius: '8px' }}>
        <h3 style={{ marginTop: 0 }}>Lubrication Regimes (Nigerian Tropical Conditions)</h3>
        <div style={{ display: 'grid', gridTemplateColumns: '1fr 1fr 1fr', gap: '15px' }}>
          <div>
            <h4 style={{ color: '#dc2626', marginBottom: '8px' }}>Boundary (ω &lt; 100 rad/s)</h4>
            <p style={{ margin: 0, fontSize: '14px' }}>μ = 0.19 × (1 + wear_factor)</p>
            <p style={{ margin: '5px 0 0 0', fontSize: '12px', color: '#666' }}>
              Metal-to-metal contact dominates. High friction due to surface asperities. 
              Nigerian conditions: 27% higher than temperate baseline (0.15).
            </p>
          </div>
          <div>
            <h4 style={{ color: '#ea580c', marginBottom: '8px' }}>Mixed (100 ≤ ω &lt; 1000 rad/s)</h4>
            <p style={{ margin: 0, fontSize: '14px' }}>μ = 0.11 × (1 + 0.5 × wear_factor)</p>
            <p style={{ margin: '5px 0 0 0', fontSize: '12px', color: '#666' }}>
              Partial fluid film with some asperity contact. Transition regime.
              Nigerian conditions: 38% higher than temperate baseline (0.08).
            </p>
          </div>
          <div>
            <h4 style={{ color: '#16a34a', marginBottom: '8px' }}>Hydrodynamic (ω ≥ 1000 rad/s)</h4>
            <p style={{ margin: 0, fontSize: '14px' }}>μ = 0.045 × (1 + 0.2 × wear_factor)</p>
            <p style={{ margin: '5px 0 0 0', fontSize: '12px', color: '#666' }}>
              Complete fluid film separation. Lowest friction. Viscous shearing of lubricant.
              Nigerian conditions: 50% higher than temperate baseline (0.03).
            </p>
          </div>
        </div>
        
        <div style={{ marginTop: '15px', padding: '10px', backgroundColor: 'white', borderRadius: '4px' }}>
          <strong>Key Observations:</strong>
          <ul style={{ margin: '8px 0', paddingLeft: '20px', fontSize: '14px' }}>
            <li>Higher wear factors increase friction across all regimes</li>
            <li>Friction decreases with increasing velocity (classic Stribeck behavior)</li>
            <li>Tropical conditions elevate baseline friction by 27-50%</li>
            <li>Boundary regime most affected by wear (multiplied by 1+wear)</li>
            <li>Hydrodynamic regime least affected by wear (multiplied by 1+0.2×wear)</li>
          </ul>
        </div>
      </div>
    </div>
  );
};

export default StribeckCurve;
```
