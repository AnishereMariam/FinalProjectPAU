```python
# Cell 1: Load the saved data
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split

# Load data
df = pd.read_csv('../data/ai4i2020_local.csv')
print(f"Data loaded: {df.shape}")
```

```python
# Cell 2: Prepare features and target

# Define feature columns
feature_cols = ['Air temperature [K]', 'Process temperature [K]', 
                'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']

# Encode categorical 'Type' column
type_mapping = {'L': 0, 'M': 1, 'H': 2}
df['Type_encoded'] = df['Type'].map(type_mapping)
feature_cols.append('Type_encoded')

# Extract features and target
X = df[feature_cols]
y = df['Machine failure']

print(f"Features shape: {X.shape}")
print(f"Target shape: {y.shape}")
print(f"Features: {feature_cols}")
```

```python
# Cell 3: Handle missing values and scale features
# Check for missing values
print("Missing values:")
print(X.isnull().sum())

# Scale features (important for neural networks)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for easier handling
X_scaled = pd.DataFrame(X_scaled, columns=feature_cols)

print("Features scaled successfully!")
print(X_scaled.describe())
```

```python
# Cell 4: Split the data
# Split into train (70%), validation (15%), and test (15%)
X_temp, X_test, y_temp, y_test = train_test_split(
    X_scaled, y, test_size=0.15, random_state=42, stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.176, random_state=42, stratify=y_temp
)  # 0.176 * 0.85 ≈ 0.15

print(f"Training set: {X_train.shape}")
print(f"Validation set: {X_val.shape}")
print(f"Test set: {X_test.shape}")

# Check failure distribution in each set
print(f"\nFailure rates:")
print(f"Train: {y_train.mean():.2%}")
print(f"Val: {y_val.mean():.2%}")
print(f"Test: {y_test.mean():.2%}")
```

```python
# Cell 5: Save preprocessed data
import pickle

# Create a dictionary with all processed data
data_dict = {
    'X_train': X_train,
    'X_val': X_val,
    'X_test': X_test,
    'y_train': y_train,
    'y_val': y_val,
    'y_test': y_test,
    'scaler': scaler,
    'feature_cols': feature_cols
}

# Save to pickle file
with open('../data/preprocessed_data.pkl', 'wb') as f:
    pickle.dump(data_dict, f)

print("Preprocessed data saved to ../data/preprocessed_data.pkl")
```
